```python
import os
import pandas as pd
import sqlite3

# Path to your CSV file
csv_path = r"C:\Users\ginas\OneDrive\Documents\Retail-SQL-Analysis\data\retail_dataset_5000 (2).csv"


# Load CSV
df = pd.read_csv(csv_path)

# Path where the SQLite database will be created
db_path = r"C:\Users\ginas\OneDrive\Documents\Retail-SQL-Analysis\data\retail_sales.db"

# If a file already exists at db_path and is not a valid database, delete it
if os.path.exists(db_path):
    os.remove(db_path)

# Connect to SQLite (this creates a new database file)
conn = sqlite3.connect(db_path)

# Write the CSV data into a table named "sales"
df.to_sql("sales", conn, if_exists="replace", index=False)

# Close connection
conn.close()

print("SQLite database created successfully at:", db_path)
```
